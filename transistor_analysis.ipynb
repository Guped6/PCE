# Analyse des paramètres de transistors CMOS

Ce notebook calcule les paramètres technologiques des NMOS et PMOS.

```python
import numpy as np
import pandas as pd
```

Paramètres de calculs

```python
# constantes physiques

epsilon_0 = 8.854e-12
epsilon_ox_rel = 3.9
epsilon_ox = epsilon_0 * epsilon_ox_rel

q = 1.602e-19

# =========================
# Paramètres NMOS
# =========================

mu_n = 0.05

tox_n = 5e-9

VA_n = 20

Lov_n = 50e-9

Rsheet_n = 100

L_n = 180e-9

W_n = 10e-6

Vfb_n = -0.95

phi_n = 0.3

gamma_n = 0.4


# =========================
# Paramètres PMOS
# =========================

mu_p = 0.02

tox_p = 5e-9

VA_p = 15

Lov_p = 50e-9

Rsheet_p = 120

L_p = 180e-9

W_p = 10e-6

Vfb_p = -0.9

phi_p = 0.3

gamma_p = 0.45
```

Calculs sous forme de fonctions

```python
# =========================
# 1. Cox
# =========================

def calc_Cox(tox):

    return epsilon_ox/tox


# =========================
# 2. Transconductance Kp
# =========================

def calc_Kp(mu,tox):

    Cox = calc_Cox(tox)

    return mu * Cox


# =========================
# 3. Channel length modulation
# =========================

def calc_lambda(VA):

    return 1/VA


# =========================
# 4. CGSO
# =========================

def calc_CGSO(Cox,Lov):

    return Cox * Lov


# =========================
# 5. CGDO
# =========================

def calc_CGDO(Cox,Lov):

    return Cox * Lov


# =========================
# 6. CGBO
# =========================

def calc_CGBO(CGSO,factor=0.3):

    return factor * CGSO


# =========================
# 7. Drain resistance
# =========================

def calc_Rd(Rsheet,L,W):

    return Rsheet * L/W


# =========================
# 8. Source resistance
# =========================

def calc_Rs(Rsheet,L,W):

    return Rsheet * L/W


# =========================
# 9. Threshold voltage
# =========================

def calc_Vt0(Vfb,phi,gamma):

    return Vfb + 2*phi + gamma*np.sqrt(2*phi)
```

```python
# =========================
# NMOS calculations
# =========================

Cox_n = calc_Cox(tox_n)

Kp_n = calc_Kp(mu_n,tox_n)

lambda_n = calc_lambda(VA_n)

CGSO_n = calc_CGSO(Cox_n,Lov_n)

CGDO_n = calc_CGDO(Cox_n,Lov_n)

CGBO_n = calc_CGBO(CGSO_n)

Rd_n = calc_Rd(Rsheet_n,L_n,W_n)

Rs_n = calc_Rs(Rsheet_n,L_n,W_n)

Vt0_n = calc_Vt0(Vfb_n,phi_n,gamma_n)


# =========================
# PMOS calculations
# =========================

Cox_p = calc_Cox(tox_p)

Kp_p = calc_Kp(mu_p,tox_p)

lambda_p = calc_lambda(VA_p)

CGSO_p = calc_CGSO(Cox_p,Lov_p)

CGDO_p = calc_CGDO(Cox_p,Lov_p)

CGBO_p = calc_CGBO(CGSO_p)

Rd_p = calc_Rd(Rsheet_p,L_p,W_p)

Rs_p = calc_Rs(Rsheet_p,L_p,W_p)

Vt0_p = calc_Vt0(Vfb_p,phi_p,gamma_p)

```

```python
# =========================
# Final table
# =========================

technologies = {

"NMOS":[

Kp_n,
Vt0_n,
lambda_n,
CGSO_n,
CGDO_n,
CGBO_n,
Rd_n,
Rs_n

],

"PMOS":[

Kp_p,
Vt0_p,
lambda_p,
CGSO_p,
CGDO_p,
CGBO_p,
Rd_p,
Rs_p

]

}

index = [

"Kp",
"Vt0",
"Lambda",
"CGSO",
"CGDO",
"CGBO",
"Rd",
"Rs"

]

df = pd.DataFrame(technologies,index=index)

df = df.transpose()

df

df.style.format({

"Kp":"{:.2e}",
"Lambda":"{:.3f}",
"CGSO":"{:.2e}",
"CGDO":"{:.2e}",
"CGBO":"{:.2e}",
"Rd":"{:.2f}",
"Rs":"{:.2f}"

})
```
